```python
import os
import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from PIL import Image, ImageFile

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
print("TensorFlow version:", tf.__version__)
print("GPU devices:", tf.config.list_physical_devices("GPU"))
```

```python
print("Current working directory:")
print(Path.cwd())
print("\nFiles and folders here:")
for item in Path.cwd().iterdir():
    print(item.name)
```

```python
DATASET_DIR = Path("DATASET")

print("Dataset path:", DATASET_DIR.resolve())
print("Path exists:", DATASET_DIR.exists())

if DATASET_DIR.exists():
    print("\nSubfolders:")
    for folder in DATASET_DIR.iterdir():
        if folder.is_dir():
            print(folder.name)
```

```python
IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

filepaths = []
labels = []

for class_folder in sorted(DATASET_DIR.iterdir()):
    if class_folder.is_dir():
        for image_path in class_folder.rglob("*"):
            if image_path.is_file() and image_path.suffix.lower() in IMAGE_EXTENSIONS:
                filepaths.append(str(image_path))
                labels.append(class_folder.name)

image_df = pd.DataFrame({
    "Filepath": filepaths,
    "Label": labels
})

print("Total images found:", len(image_df))
print("\nImages per class:")
print(image_df["Label"].value_counts().sort_index())

image_df.head()
```

```python
# Let's see if some flowers have way more images than others

class_counts = image_df["Label"].value_counts().sort_index()

print("Images in each class:")
print(class_counts)

# A graph makes the difference easier to see
plt.figure(figsize=(9, 5))
class_counts.plot(kind="bar")

plt.title("Number of Images per Flower Class")
plt.xlabel("Flower")
plt.ylabel("Number of Images")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Let's randomly look at 4 images from each flower class
# This helps us spot wrong labels, blurry photos or irrelevant images

sample_df = image_df.groupby("Label", group_keys=False).sample(
    n=4,
    random_state=42
)

flower_classes = sorted(image_df["Label"].unique())

fig, axes = plt.subplots(
    nrows=len(flower_classes),
    ncols=4,
    figsize=(12, 15)
)

for row, flower in enumerate(flower_classes):
    flower_images = sample_df[sample_df["Label"] == flower]

    for col in range(4):
        img_path = flower_images.iloc[col]["Filepath"]

        # Convert to RGB so the images display consistently
        with Image.open(img_path) as img:
            axes[row, col].imshow(img.convert("RGB"))

        axes[row, col].set_title(flower if col == 0 else "")
        axes[row, col].axis("off")

plt.tight_layout()
plt.show()
```

```python
# The dataset has images with different dimensions.
# Let's find out what sizes we're actually dealing with.

image_sizes = []

for path in image_df["Filepath"]:
    try:
        with Image.open(path) as img:
            image_sizes.append(img.size)  # (width, height)
    except Exception:
        image_sizes.append(None)

image_df["Image_Size"] = image_sizes

print("Most common image sizes:")
print(image_df["Image_Size"].value_counts().head(10))
```

```python
# Some files might have a .jpg extension but still be unreadable.
# Let's try opening every image to check.

corrupted_images = []

for path in image_df["Filepath"]:
    try:
        with Image.open(path) as img:
            img.verify()
    except Exception:
        corrupted_images.append(path)

print("Total images checked:", len(image_df))
print("Corrupted images found:", len(corrupted_images))
```

```python
import hashlib

# A hash helps us identify files with exactly the same contents
image_hashes = []

for path in image_df["Filepath"]:
    with open(path, "rb") as file:
        image_hashes.append(hashlib.md5(file.read()).hexdigest())

image_df["Image_Hash"] = image_hashes

# Find repeated hashes
duplicate_mask = image_df.duplicated(
    subset="Image_Hash",
    keep=False
)

duplicates = image_df[duplicate_mask].sort_values("Image_Hash")

print("Images belonging to duplicate groups:", len(duplicates))
print("Unique duplicate groups:", duplicates["Image_Hash"].nunique())

display(duplicates[["Filepath", "Label", "Image_Hash"]].head(20))
```

```python
# Let's see whether images are RGB, grayscale, or use other colour modes

image_modes = []

for path in image_df["Filepath"]:
    with Image.open(path) as img:
        image_modes.append(img.mode)

image_df["Image_Mode"] = image_modes

print("Image colour modes:")
print(image_df["Image_Mode"].value_counts())
```

```python
# We saw that images have different sizes.
# Let's see how their widths and heights are distributed.

valid_sizes = image_df["Image_Size"].dropna()

widths = valid_sizes.apply(lambda size: size[0])
heights = valid_sizes.apply(lambda size: size[1])

plt.figure(figsize=(8, 5))
plt.scatter(widths, heights, alpha=0.5)

plt.title("Image Width vs Height")
plt.xlabel("Width (pixels)")
plt.ylabel("Height (pixels)")
plt.tight_layout()
plt.show()

print("Smallest width:", widths.min())
print("Largest width:", widths.max())
print("Smallest height:", heights.min())
print("Largest height:", heights.max())
```

```python
# Let's check whether identical images have different labels

duplicate_summary = (
    duplicates.groupby("Image_Hash")
    .agg(
        Number_of_Copies=("Filepath", "count"),
        Labels=("Label", lambda x: ", ".join(sorted(set(x)))),
        Filepaths=("Filepath", list)
    )
    .reset_index()
)

display(duplicate_summary)
```

```python
# Show duplicate images with their labels so we can review them

for _, row in duplicate_summary.iterrows():
    paths = row["Filepaths"]

    if len(set(row["Labels"].split(", "))) > 1:
        print("Labels:", row["Labels"])

        fig, axes = plt.subplots(1, len(paths), figsize=(4 * len(paths), 4))

        if len(paths) == 1:
            axes = [axes]

        for ax, path in zip(axes, paths):
            with Image.open(path) as img:
                ax.imshow(img.convert("RGB"))

            ax.set_title(Path(path).parent.name)
            ax.axis("off")

        plt.tight_layout()
        plt.show()
```

```python
# We'll keep the original dataset unchanged.
# This creates a cleaned list of image paths for our model.

# First, identify images whose exact duplicates have conflicting labels
conflicting_hashes = (
    duplicate_summary[
        duplicate_summary["Labels"].apply(lambda x: len(x.split(", ")) > 1)
    ]["Image_Hash"]
    .tolist()
)

# Exclude all copies of these ambiguous images
clean_df = image_df[
    ~image_df["Image_Hash"].isin(conflicting_hashes)
].copy()

# Now remove repeated copies of otherwise valid images.
# Keep just one copy of each exact image.
clean_df = clean_df.drop_duplicates(
    subset="Image_Hash",
    keep="first"
).copy()

print("Original images:", len(image_df))
print("Images after duplicate handling:", len(clean_df))
print("\nImages per class after cleaning:")
print(clean_df["Label"].value_counts().sort_index())
```

```python
# Check that no exact duplicate images remain

print("Remaining duplicate images:",
      clean_df["Image_Hash"].duplicated().sum())

print("\nRemaining class counts:")
print(clean_df["Label"].value_counts().sort_index())

# Keep the original file list intact; only the inventory has changed.
```

```python
# The model learns from training images.
# Validation helps us tune it, and test data is saved for final evaluation.

train_df, temp_df = train_test_split(
    clean_df,
    test_size=0.30,
    stratify=clean_df["Label"],
    random_state=42
)

# Split the remaining 30% equally into validation and test sets
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["Label"],
    random_state=42
)

print("Training images:", len(train_df))
print("Validation images:", len(val_df))
print("Test images:", len(test_df))

print("\nTraining class distribution:")
print(train_df["Label"].value_counts().sort_index())

print("\nValidation class distribution:")
print(val_df["Label"].value_counts().sort_index())

print("\nTest class distribution:")
print(test_df["Label"].value_counts().sort_index())
```

```python
# An image must not appear in more than one split.

train_paths = set(train_df["Filepath"])
val_paths = set(val_df["Filepath"])
test_paths = set(test_df["Filepath"])

print("Train-validation overlap:", len(train_paths & val_paths))
print("Train-test overlap:", len(train_paths & test_paths))
print("Validation-test overlap:", len(val_paths & test_paths))
```

```python
# MobileNetV2 will receive images of one consistent size
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

# Keep the class order fixed so predictions map to the right flower
CLASS_NAMES = sorted(clean_df["Label"].unique())

print("Image size:", IMG_SIZE)
print("Batch size:", BATCH_SIZE)
print("Class names:", CLASS_NAMES)
```

```python
# Resize images and prepare them for MobileNetV2
def load_and_preprocess_image(filepath, label):
    image = tf.io.read_file(filepath)
    image = tf.io.decode_image(image, channels=3, expand_animations=False)
    image.set_shape([None, None, 3])
    image = tf.image.resize(image, IMG_SIZE)
    image = tf.keras.applications.mobilenet_v2.preprocess_input(image)
    return image, label
```

```python
# Convert flower names into numbers for the model
label_to_index = {name: index for index, name in enumerate(CLASS_NAMES)}

def create_dataset(dataframe, training=False):
    dataset = tf.data.Dataset.from_tensor_slices((
        dataframe["Filepath"].to_numpy(),
        dataframe["Label"].map(label_to_index).to_numpy()
    ))

    if training:
        dataset = dataset.shuffle(len(dataframe), seed=42)

    dataset = dataset.map(
        load_and_preprocess_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )
    dataset = dataset.batch(BATCH_SIZE)
    dataset = dataset.prefetch(tf.data.AUTOTUNE)

    return dataset


train_ds = create_dataset(train_df, training=True)
val_ds = create_dataset(val_df)
test_ds = create_dataset(test_df)
```

```python
images, labels = next(iter(train_ds))

print("Image batch shape:", images.shape)
print("Label batch shape:", labels.shape)
print("Pixel value range:",
      float(tf.reduce_min(images)),
      "to",
      float(tf.reduce_max(images)))
```

```python
# Make small changes to training images so the model doesn't memorise them
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.1),
    tf.keras.layers.RandomZoom(0.1),
], name="data_augmentation")
```

```python
# Load MobileNetV2 with pretrained ImageNet weights
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)
# Freeze the pretrained layers so we keep their learned features
base_model.trainable = False

print("MobileNetV2 loaded successfully!")
print("Pretrained layers frozen:", not base_model.trainable)
```

```python
# Build the model that will classify our five flower types
inputs = tf.keras.Input(shape=(224, 224, 3))

x = data_augmentation(inputs)
x = base_model(x, training=False)
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dropout(0.2)(x)

outputs = tf.keras.layers.Dense(
    len(CLASS_NAMES),
    activation="softmax"
)(x)

model = tf.keras.Model(inputs, outputs)

model.summary()
```

```python
# Set up the model for training
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully!")
```

```python
# Save the best model and stop training if it stops improving
callbacks = [
    tf.keras.callbacks.ModelCheckpoint(
        "best_flower_model.keras",
        monitor="val_loss",
        save_best_only=True
    ),
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True
    )
]
```

```python
# Train the model using our training data and check validation performance
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    callbacks=callbacks
)
```

```python
# Plot accuracy and loss during training
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("Model Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Evaluate the final model on unseen test images
test_loss, test_accuracy = model.evaluate(test_ds)

print("Test Accuracy:", round(test_accuracy * 100, 2), "%")
print("Test Loss:", round(test_loss, 4))
```

```python
# Check predictions for each flower class
y_true = test_df["Label"].map(label_to_index).to_numpy()
y_pred = np.argmax(model.predict(test_ds), axis=1)

print("Classification Report:\n")
print(classification_report(
    y_true,
    y_pred,
    labels=range(len(CLASS_NAMES)),
    target_names=CLASS_NAMES,
    zero_division=0
))

# Show which flower classes the model confuses
cm = confusion_matrix(y_true, y_pred, labels=range(len(CLASS_NAMES)))

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=CLASS_NAMES
).plot(cmap="Blues", xticks_rotation=45)

plt.title("Flower Classification Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
# Save the trained model for future use
model.save("flower_model.keras")

print("Model saved successfully!")
```

```python
# Pick one image from the test dataset and predict its flower class
sample = test_df.iloc[0]

image = tf.keras.utils.load_img(
    sample["Filepath"],
    target_size=IMG_SIZE
)

image_array = tf.keras.utils.img_to_array(image)
image_array = tf.keras.applications.mobilenet_v2.preprocess_input(image_array)
image_array = np.expand_dims(image_array, axis=0)

prediction = model.predict(image_array, verbose=0)[0]
predicted_flower = CLASS_NAMES[np.argmax(prediction)]

plt.imshow(image)
plt.title(f"Actual: {sample['Label']} | Predicted: {predicted_flower}")
plt.axis("off")
plt.show()

print("Predicted flower:", predicted_flower)
print("Confidence:", round(float(np.max(prediction)) * 100, 2), "%")
```
